```python
import math

# Entradas de teste
P = 320.0  # Potência ativa em kW
FP_atual = 0.80  # Fator de potência atual
FP_alvo = 0.92  # Fator de potência alvo regulatório

if FP_atual < FP_alvo:
    # Cálculo dos ângulos (em radianos) baseados nos fatores de potência
    phi_atual = math.acos(FP_atual)
    phi_alvo = math.acos(FP_alvo)

    # Cálculo da potência reativa capacitiva Qc = P * (tan(phi_atual) - tan(phi_alvo))
    Q_c = P * (math.tan(phi_atual) - math.tan(phi_alvo))

    print(f"Potência Reativa Requerida (Qc): {Q_c:.2f} kVAr")
else:
    print(f"O fator de potência atual ({FP_atual:.2f}) já atende ou supera o limite regulatório de {FP_alvo:.2f}.")
```
