```python
# Dados de entrada (Teste)
P_saida_kw = 450.0
P_nucleo_w = 1200.0
P_cobre_w = 3800.0

# Conversão das perdas de Watts (W) para Quilowatts (kW)
P_nucleo_kw = P_nucleo_w / 1000.0
P_cobre_kw = P_cobre_w / 1000.0

# Cálculo da potência de entrada: P_entrada = P_saída + P_núcleo + P_cobre
P_entrada_kw = P_saida_kw + P_nucleo_kw + P_cobre_kw

# Cálculo do rendimento percentual (η)
rendimento = (P_saida_kw / P_entrada_kw) * 100.0

# Exibição dos resultados com formatação de três casas decimais
print(f"Potência de Entrada: {P_entrada_kw:.3f} kW | Rendimento: {rendimento:.3f}%")
```
