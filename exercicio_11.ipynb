```python
# Exercício 11: Varredura de Telemetria com Detecção de Falta (break) e Filtragem de Ruído (continue)

# Vetor de correntes (A)
correntes = [82.0, 85.5, -99.0, 90.0, 94.2, -50.0, 105.0, 218.0, 95.0, 91.0]

soma_validas = 0.0
contagem_validas = 0

for i, valor in enumerate(correntes):
    if valor < 0.0:
        print(f"Amostra {i} descartada: Ruído de medição detectado ({valor} A)")
        continue

    if valor > 200.0:
        print(f"\nALARME DE FALTA: Corrente de curto-circuito detectada na Amostra {i} ({valor} A)! Interrompendo monitoramento.")
        break

    # Valores válidos acumulados
    soma_validas += valor
    contagem_validas += 1
    print(f"Amostra {i} processada com sucesso: {valor} A")

# Cálculo da média das amostras válidas processadas antes do break
if contagem_validas > 0:
    media_validas = soma_validas / contagem_validas
    print(f"\nMonitoramento encerrado.")
    print(f"Leituras válidas processadas: {contagem_validas} | Média Calculada: {media_validas:.2f} A")
else:
    print("\nNenhuma leitura válida processada.")
```
