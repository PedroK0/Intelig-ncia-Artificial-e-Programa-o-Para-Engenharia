```python
# Exercício 10: Integração de Consumo Ativo Horário com Tarifação Horosazonal

# Leituras horárias de potência média consumida (kW)
demanda_24h = [
    40, 38, 35, 36, 42, 55, 70, 85, 90, 92, 88, 86,
    84, 87, 91, 95, 110, 135, 150, 145, 130, 95, 65, 48
]

# Tarifas (R$/kWh)
tarifa_fora_ponta = 0.65
tarifa_ponta = 1.85

# Índices correspondentes ao Horário de Ponta
indices_ponta = {18, 19, 20}

consumo_total_kwh = 0.0
custo_total_reais = 0.0

# Processamento de cada hora do dia (delta_t = 1 h)
for hora, potencia in enumerate(demanda_24h):
    energia_hora = potencia * 1.0  # E = P * delta_t
    consumo_total_kwh += energia_hora

    if hora in indices_ponta:
        custo_total_reais += energia_hora * tarifa_ponta
    else:
        custo_total_reais += energia_hora * tarifa_fora_ponta

# Exibição do Resultado
print(f"Consumo Total: {consumo_total_kwh:.2f} kWh | Custo Total: R$ {custo_total_reais:.2f}")
```
