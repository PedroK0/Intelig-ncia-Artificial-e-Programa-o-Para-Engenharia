```python
# Exercício 14: Implementação de Buffer Circular Oscilográfico FIFO

# Buffer oscilográfico inicial (capacidade máxima de 5 amostras)
buffer = [81.2, 82.0, 80.9, 83.5, 82.8]

# Novas amostras provenientes do conversor A/D
novas_leituras = [86.4, 91.0, 94.5]

print(f"Buffer Inicial: {buffer}")

# Simulação do processamento FIFO ciclo a ciclo
for nova_amostra in novas_leituras:
    buffer.append(nova_amostra)    # Insere o dado mais recente no fim do buffer
    buffer.pop(0)                 # Remove o dado mais antigo do início do buffer
    print(f"Nova leitura: {nova_amostra:4.1f} A | Estado do Buffer: {buffer}")

print(f"\nBuffer Final Consolidado: {buffer}")
```
