```markdown
# Exercício 15: Processamento Vetorial com List Comprehensions (Norma PRODIST)

Nesta apresentação, demonstraremos como utilizar compreensões de lista (*list comprehensions*) para realizar processamento vetorial rápido e eficiente em dados de telemetria de tensão, atendendo aos limites regulatórios do PRODIST.
```

````python
# 1. Definição dos dados de entrada (leituras de tensão) e da tensão base
leituras_v = [220.0, 204.5, 218.0, 235.2, 222.1, 198.0, 228.4, 232.0, 208.5]
V_base = 220.0

# 2. Conversão para o sistema por unidade (pu) utilizando List Comprehension
leituras_pu = [round(v / V_base, 3) for v in leituras_v]

# 3. Filtragem das tensões transgressoras segundo o PRODIST (< 209V ou > 231V)
transgressoes = [v for v in leituras_v if v < 209.0 or v > 231.0]

# Exibição dos resultados
print(f"Leituras originais (V): {leituras_v}")
print(f"Leituras em pu (Base {V_base}V): {leituras_pu}")
print(f"Leituras transgressoras (PRODIST): {transgressoes}")
````

```markdown
## Explicação do Código para a Apresentação

* **`leituras_v = [...]` e `V_base = 220.0`**: Definimos a lista contendo as medições de tensão coletadas no barramento e a tensão base de referência para o cálculo do sistema por unidade (p.u.).
* **`leituras_pu = [round(v / V_base, 3) for v in leituras_v]`**:
  * Uma *list comprehension* que percorre cada leitura `v` na lista `leituras_v`.
  * Divide cada valor pela `V_base` para obter o valor em p.u.
  * Aplica a função `round(..., 3)` para limitar o resultado a exatamente 3 casas decimais, conforme solicitado.
* **`transgressoes = [v for v in leituras_v if v < 209.0 or v > 231.0]`**:
  * Outra *list comprehension* que atua como um filtro analítico rápido.
  * Ela avalia cada tensão `v` e só a adiciona à nova lista se a condição lógica do PRODIST for verdadeira: se for menor que `209.0 V` ou maior que `231.0 V`.
* **`print(...)`**: Exibe de forma clara os resultados processados para validação.
```
