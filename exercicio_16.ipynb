# Exercício 16: Proteção e Imutabilidade de Ajustes IED com Tuplas

Este notebook apresenta a implementação de segurança com tuplas para dados de placa de transformadores e o tratamento de fasores polares utilizando empacotamento.

```python
import math

# 1. Definição da tupla de dados de placa do transformador (V1, V2, S_nom, Z%)
trafo_placa = (13800.0, 220.0, 500.0, 4.5)
print(f"[CONFIG] Parâmetros de Placa carregados: {trafo_placa}\n")

# 2. Tentativa de alteração acidental da potência nominal (índice 2) para 750.0 kVA
try:
    # Como tuplas são imutáveis, esta linha deve levantar um TypeError
    trafo_placa[2] = 750.0
except TypeError as erro:
    print("[ALERTA DE SEGURANÇA] Bloqueio de sobreescrita de placa ativado com sucesso!")
    print(f"[DETALHES DO ERRO] {erro}\n")

# 3. Função que converte fasor retangular (I_r, I_i) para polar (magnitude, angulo em graus)
def converter_para_polar(I_r, I_i):
    magnitude = math.sqrt(I_r**2 + I_i**2)
    angulo_rad = math.atan2(I_i, I_r)
    angulo_graus = math.degrees(angulo_rad)
    return (magnitude, angulo_graus) # Retorna uma tupla empacotada

# 4. Exemplo de uso: Corrente retangular (4.0 A, 3.0 A)
fasor_polar = converter_para_polar(4.0, 3.0)

# 5. Desempacotamento do resultado em duas variáveis isoladas
I_mag, I_ang = fasor_polar

# Exibição dos resultados finais
print(f"Fasor polar completo (empacotado): {fasor_polar}")
print(f"Corrente (Desempacotada) -> Magnitude: {I_mag:.2f} A | Ângulo: {I_ang:.2f}°")
```

## Explicação do Código para a Apresentação

* **`trafo_placa = (13800.0, 220.0, 500.0, 4.5)`**: Definimos os dados nominais fixados de fábrica utilizando uma tupla `(...)`. Na engenharia de software, tuplas garantem segurança pois são coleções imutáveis.
* **`try: trafo_placa[2] = 750.0 except TypeError ...`**: Simulamos um erro em tempo de execução onde alguém tenta alterar o terceiro parâmetro (potência de 500 kVA para 750 kVA). O Python barra a operação e dispara um `TypeError`, o qual capturamos de forma controlada para logar um alerta de segurança.
* **`converter_para_polar(I_r, I_i)`**: Uma função que utiliza álgebra trigonométrica básica para converter a parte real e imaginária da corrente elétrica em coordenadas polares, retornando o resultado em uma nova tupla empacotada de retorno.
* **`I_mag, I_ang = fasor_polar`**: Realiza o desempacotamento direto da tupla retornada, distribuindo os valores de magnitude e ângulo para variáveis individuais de forma limpa e idiomática.
