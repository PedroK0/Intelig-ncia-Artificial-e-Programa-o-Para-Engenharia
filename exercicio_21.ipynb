```markdown
# Exercício 21: Função Modular com Type Hinting e Docstring Normativa (PEP 484)

Nesta seção, estruturamos uma função robusta e tipada para cálculo da queda de tensão trifásica em redes de distribuição de energia elétrica ($13,8\text{ kV}$), em total conformidade com as recomendações de engenharia de software aplicadas à infraestrutura (PEP 484 e docstrings normativas).
```

````python
import math

def queda_tensao_trifasica(
    i_linha: float,
    comprimento_km: float,
    fator_potencia: float,
    r_km: float = 0.19,
    x_km: float = 0.08
) -> tuple[float, float, bool]:
    """
    Calcula a queda de tensão e a regulação de tensão de uma linha trifásica de 13,8 kV.

    Fórmula:
        Delta_V = sqrt(3) * I * [R_km * FP + X_km * sen(arccos(FP))] * L
        Regulacao (%) = (Delta_V / V_base) * 100%

    Parâmetros:
        i_linha (float): Corrente de linha que percorre o condutor, em Ampères (A).
        comprimento_km (float): Comprimento da linha de transmissão/distribuição, em km.
        fator_potencia (float): Fator de potência indutivo da carga (adimensional).
        r_km (float, opcional): Resistência linear do condutor, em Ohms/km (padrão: 0.19).
        x_km (float, opcional): Reatância linear do condutor, em Ohms/km (padrão: 0.08).

    Retorna:
        tuple[float, float, bool]:
            - delta_v (float): Queda de tensão aproximada em Volts (V).
            - regulacao_pct (float): Regulação percentual de tensão na base de 13800 V.
            - conforme (bool): Indicativo se a regulação é <= 5.0% (conforme PRODIST).
    """
    # 1. Obtenção do ângulo correspondente ao fator de potência e cálculo do seno correspondente
    angulo_rad = math.acos(fator_potencia)
    seno_phi = math.sin(angulo_rad)

    # 2. Aplicação da fórmula de queda de tensão aproximada trifásica
    impedancia_equivalente = (r_km * fator_potencia) + (x_km * seno_phi)
    delta_v = math.sqrt(3) * i_linha * impedancia_equivalente * comprimento_km

    # 3. Cálculo da regulação de tensão percentual (Tensão base = 13800 V)
    V_base = 13800.0
    regulacao_pct = (delta_v / V_base) * 100.0

    # 4. Avaliação normativa de conformidade (limite regulatório de 5%)
    conforme = regulacao_pct <= 5.0

    return delta_v, regulacao_pct, conforme

# --- Teste de Validação Operativa ---
I_teste = 120.0
L_teste = 2.5
FP_teste = 0.88

dv, reg, status = queda_tensao_trifasica(I_teste, L_teste, FP_teste)

print("=== RELATÓRIO DO CÁLCULO DE QUEDA DE TENSÃO (13,8 kV) ===\n")
print(f"Corrente do Circuito: {I_teste:.1f} A")
print(f"Comprimento da Rede : {L_teste:.1f} km")
print(f"Fator de Potência   : {FP_teste:.2f}\n")
print(f"-> Queda de Tensão (Delta V) : {dv:.2f} V")
print(f"-> Regulação de Tensão       : {reg:.2f}%")
print(f"-> Situação Regulatória      : {'CONFORME (Adequado)' if status else 'NÃO CONFORME (Crítico)'}")
````

```markdown
* **Assinatura com Type Hinting (`-> tuple[...]`)**: Especifica estritamente os tipos esperados para os parâmetros (`float`) e o tipo do retorno, ajudando ferramentas de análise estática e IDEs a capturarem incompatibilidades de tipo em tempo de desenvolvimento.
```
