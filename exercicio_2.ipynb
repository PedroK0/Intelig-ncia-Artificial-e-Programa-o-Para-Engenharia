```python
import math

# Entradas de teste
V_r = 127.0
V_i = -73.32

# Cálculo da magnitude eficaz |V| = √(V_r² + V_i²)
magnitude = math.sqrt(V_r**2 + V_i**2)

# Cálculo do ângulo em graus θ = atan2(V_i, V_r) * (180 / π)
angulo_rad = math.atan2(V_i, V_r)
angulo_graus = math.degrees(angulo_rad)

# Exibição do resultado formatado
print(f"Fasor Polar: |V| = {magnitude:.2f} V | Ângulo = {angulo_graus:.2f}°")
```
