```python
def validar_telemetria(V, I, f):
    # Validações físicas do sinal
    if V >= 0.0 and I >= 0.0 and 45.0 <= f <= 65.0:
        return "TELEMETRIA_VALIDA"
    else:
        return "FALHA_SINAL: Transdutor corrompido"

# Teste 1
V1, I1, f1 = 220.0, -5.0, 60.0
print(f"Entrada 1 -> V: {V1}V, I: {I1}A, f: {f1}Hz")
print(f"Saída 1: {validar_telemetria(V1, I1, f1)}\n")

# Teste 2
V2, I2, f2 = 13800.0, 48.2, 59.95
print(f"Entrada 2 -> V: {V2}V, I: {I2}A, f: {f2}Hz")
print(f"Saída 2: {validar_telemetria(V2, I2, f2)}")
```
