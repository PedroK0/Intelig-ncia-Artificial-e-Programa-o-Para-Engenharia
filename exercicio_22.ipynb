```markdown
# Exercício 22: Operações Fasoriais e Lei de Ohm com Vetores Complexos no NumPy

Nesta seção, realizamos cálculos de regime permanente em circuitos trifásicos desbalanceados. Utilizaremos o poder de processamento vetorial do **NumPy** com o tipo `np.complex128` para aplicar a Lei de Ohm fasorial ($Ġ = \dot{V} / \dot{Z}$) simultaneamente nas três fases de forma rápida e otimizada.
```

````python
import numpy as np

# 1. Definição dos fasores de tensão de fase utilizando a forma de Euler:
# V = V_eficaz * e^(j * theta_rad)
tensoes_ang_graus = np.array([0.0, -120.0, 120.0])
tensoes_ang_rad = np.radians(tensoes_ang_graus)

V_dot = 220.0 * np.exp(1j * tensoes_ang_rad)

# 2. Definição das impedâncias complexas de carga desbalanceadas
Z_dot = np.array([
    10.0 + 2j,   # Z_A
    12.0 + 5j,   # Z_B
    8.0 - 3j     # Z_C
], dtype=np.complex128)

# 3. Aplicação simultânea da Lei de Ohm Fasorial (Vetorização NumPy)
I_dot = V_dot / Z_dot

# 4. Extração de Módulos (np.abs) e Ângulos em graus (np.angle)
modulos_corrente = np.abs(I_dot)
angulos_corrente = np.angle(I_dot, deg=True)

# --- Exibição formatada dos resultados ---
print("=== ANÁLISE FASORIAL TRIFÁSICA VIA NUMPY ===\n")

fases = ['A', 'B', 'C']
for i, fase in enumerate(fases):
    print(f"Fase {fase}:\n"
          f"  Tensão    : {V_dot[i]:.2f} V\n"
          f"  Impedância: {Z_dot[i]} Ohms\n"
          f"  Corrente  : {modulos_corrente[i]:.2f} ∠ {angulos_corrente[i]:.1f}° A\n")

print("=== VETORES EXTRAÍDOS PARA SAÍDA ESPERADA ===")
print(f"Módulos: [" + ", ".join(f"{m:.2f}" for m in modulos_corrente) + "] A")
print(f"Ângulos: [" + ", ".join(f"{a:.1f}°" for a in angulos_corrente) + "]")
````

```markdown
## Explicação do Código para a Apresentação

* **`np.exp(1j * np.radians(angulo))`**: Implementação direta da fórmula de Euler ($e^{j\theta}$), onde o sufixo `1j` define a constante imaginária no Python. A conversão de graus para radianos é essencial antes de aplicar as funções exponenciais e trigonométricas.
* **`np.complex128`**: Garante dupla precisão numérica de 64 bits para as componentes real e imaginária de cada elemento elétrico.
* **`I_dot = V_dot / Z_dot`**: Em vez de iterar sobre as fases individualmente, a Lei de Ohm é processada instantaneamente de forma vetorial. O NumPy lida internamente com a divisão complexa element-wise.
* **`np.abs(I_dot)` e `np.angle(I_dot, deg=True)`**: Extraem, respectivamente, o módulo polar (módulo das correntes) e os ângulos de fase convertidos diretamente para graus centígrados (através do parâmetro `deg=True`).
```
