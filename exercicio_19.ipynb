```markdown
# Exercício 19: Serialização de Pacotes Telemétricos em Formato JSON (MQTT/IoT)

Nesta seção, convertemos a estrutura de dados em memória referente aos alimentadores industriais para o formato de serialização JSON, amplamente utilizado em plataformas de supervisão e telemetria IoT (MQTT).
```

````python
import json

# Como o Exercício 18 já foi executado no kernel, a variável 'alimentadores' está ativa em memória.
# No entanto, para garantir a independência da célula de execução, vamos recriar o conjunto final:
pacote_alimentadores = [
    {
        "nome": "Alimentador 1",
        "correntes": [115.0, 114.0, 116.0],
        "i_medio": 115.0,
        "desbalanceamento_pct": 0.87,
        "status": "CONFORME"
    },
    {
        "nome": "Alimentador 2",
        "correntes": [140.0, 138.0, 142.0],
        "i_medio": 140.0,
        "desbalanceamento_pct": 1.43,
        "status": "CONFORME"
    },
    {
        "nome": "Alimentador 3",
        "correntes": [90.0, 125.0, 88.0],
        "i_medio": 101.0,
        "desbalanceamento_pct": 23.76,
        "status": "ALERTA_DESBALANCEAMENTO"
    }
]

# 1. Serialização em JSON com formatação estruturada
# 'indent=2' para organização visual em blocos legíveis
# 'ensure_ascii=False' garante que caracteres latinos (como acentos e o símbolo 'Ó') sejam gravados corretamente
pacote_json_str = json.dumps(pacote_alimentadores, indent=2, ensure_ascii=False)

# 2. Exibição da String JSON serializada, pronta para envio telemétrico via MQTT
print("=== STRING JSON SERIALIZADA PARA TRANSMISSÃO IoT ===\n")
print(pacote_json_str)

# 3. Verificação de integridade: Testando a desserialização de volta ao dicionário original
try:
    dados_validados = json.loads(pacote_json_str)
    print("\n[VERIFICAÇÃO] Integridade sintática confirmada com sucesso! JSON válido para transferência.")
except json.JSONDecodeError as e:
    print(f"\n[ERRO] Falha na integridade do arquivo gerado: {e}")
````

```markdown
## Explicação do Código para a Apresentação

* **`import json`**: Importa o módulo nativo do Python especializado em manipular a codificação, transmissão e decodificação do padrão JSON.
* **`json.dumps(pacote_alimentadores, indent=2, ensure_ascii=False)`**:
  * O método `dumps` serializa o objeto estruturado Python para uma string textual no formato JSON.
  * **`indent=2`**: Insere espaçamento e quebras de linha para tornar o documento humanamente legível.
  * **`ensure_ascii=False`**: Evita codificar caracteres especiais latinos para sequências de escape Unicode (como `\u00d3` para 'Ó'), mantendo o texto legível e em conformidade regional (como nos nomes dos status e das variáveis).
* **`json.loads(pacote_json_str)`**: Reconverte a string JSON de volta a dicionários nativos. Executar essa instrução com sucesso comprova que a sintaxe gerada é 100% íntegra e não contém falhas estruturais.
```
