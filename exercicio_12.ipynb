```python
# Exercício 12: Simulação Transiente de Resfriamento de Motor com Critério de Parada (while)

# Condições iniciais e parâmetros físicos
T_0 = 95.0       # Temperatura inicial (°C)
T_amb = 25.0     # Temperatura ambiente (°C)
k_term = 0.08    # Constante de resfriamento (min⁻¹)
limiar = 0.5     # Critério de segurança em relação à temp. ambiente (°C)

tempo = 0
T_atual = T_0

print(f"Tempo: {tempo:2d} min | Temperatura: {T_atual:6.2f} °C")

# Laço de simulação dinâmica
while (T_atual - T_amb) >= limiar:
    tempo += 1
    T_atual = T_atual - k_term * (T_atual - T_amb)
    print(f"Tempo: {tempo:2d} min | Temperatura: {T_atual:6.2f} °C")

print(f"\nMonitoramento Concluído!")
print(f"Tempo total para religamento seguro: {tempo} minutos")
print(f"Temperatura final atingida: {T_atual:.2f} °C")
```
