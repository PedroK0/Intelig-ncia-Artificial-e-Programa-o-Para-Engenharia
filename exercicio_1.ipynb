### Exercício 1: Cálculo da Resistência de Condutores com Variação Térmica

Este script calcula a resistência elétrica de uma linha de transmissão aérea de alumínio corrigida pela temperatura de operação, utilizando a Lei de Pouillet integrada ao coeficiente de variação térmica do material.

```python
def calcular_resistencia_linha(L_km, S_mm2, T_celsius):
    # Constantes do Alumínio
    rho_20 = 0.0282  # Resistividade a 20°C (Ohm * mm2 / m)
    alpha_20 = 0.0039  # Coeficiente térmico a 20°C (1/°C)

    # Conversão de quilômetros para metros
    L_m = L_km * 1000.0

    # Cálculo da resistência usando a Lei de Pouillet com correção térmica
    R = rho_20 * (L_m / S_mm2) * (1 + alpha_20 * (T_celsius - 20))
    return R

# --- Dados de Entrada (Teste) ---
L_teste = 12.5       # em km
S_teste = 240.0      # em mm²
T_teste = 45.0       # em °C

# Execução do cálculo
resistencia_total = calcular_resistencia_linha(L_teste, S_teste, T_teste)

# Exibição do resultado esperado
print(f"Dados de Entrada:")
print(f"- Comprimento da linha (L): {L_teste} km")
print(f"- Bitola nominal (S): {S_teste} mm²")
print(f"- Temperatura de operação (T): {T_teste} °C\n")
print(f"Resistência da linha a {T_teste:.1f}°C: {resistencia_total:.4f} ohms")
```
