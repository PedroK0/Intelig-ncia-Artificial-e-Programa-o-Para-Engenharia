```python
def calcular_resistencia_linha(L_km, S_mm2, T_celsius):
    # Constantes físicas do Alumínio
    rho_20 = 0.0282  # Resistividade do alumínio a 20°C (Ohm * mm² / m)
    alpha_20 = 0.0039  # Coeficiente de temperatura a 20°C (1/°C)

    # Conversão do comprimento de quilômetros (km) para metros (m)
    L_m = L_km * 1000.0

    # Cálculo da resistência elétrica usando a Lei de Pouillet com a correção de temperatura
    R = rho_20 * (L_m / S_mm2) * (1 + alpha_20 * (T_celsius - 20))
    return R

# --- Dados de Entrada (Teste) ---
L_teste = 12.5       # Comprimento da linha de teste em km
S_teste = 240.0      # Área da seção transversal (bitola) de teste em mm²
T_teste = 45.0       # Temperatura de operação de teste em °C

# Execução do cálculo chamando a função com os dados de teste
resistencia_total = calcular_resistencia_linha(L_teste, S_teste, T_teste)

# Exibição dos dados e do resultado formatado no console
print(f"Dados de Entrada:")
print(f"- Comprimento da linha (L): {L_teste} km")
print(f"- Bitola nominal (S): {S_teste} mm²")
print(f"- Temperatura de operação (T): {T_teste} °C\n")
print(f"Resistência da linha a {T_teste:.1f}°C: {resistencia_total:.4f} ohms")
```
