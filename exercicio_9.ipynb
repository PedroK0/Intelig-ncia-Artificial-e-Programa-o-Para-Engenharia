```python
# Exercício 9: Cálculo Discreto de Tensão Eficaz (V_RMS) por Quadratura
import math

# Amostras instantâneas de tensão (V)
amostras = [0.0, 110.0, 190.5, 220.0, 190.5, 110.0, 0.0, -110.0, -190.5, -220.0, -190.5, -110.0]

# Acumulador para a soma dos quadrados
soma_quadrados = 0.0
N = len(amostras)

# Laço para somar os quadrados das tensões
for v_k in amostras:
    soma_quadrados += v_k ** 2

# Cálculo do V_RMS
v_rms = math.sqrt(soma_quadrados / N)

# Exibição do Resultado
print(f"Amostras: {N} | V_RMS Calculado: {v_rms:.2f} V")
```
