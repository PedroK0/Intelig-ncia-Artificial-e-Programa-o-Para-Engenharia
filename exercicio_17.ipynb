# Exercício 17: Modelagem Semântica de Ativos e Consulta Segura com .get()

Nesta seção, demonstraremos como estruturar dados de um alimentador elétrico usando dicionários, realizar leituras tolerantes a falhas utilizando o método `.get()` e atualizar valores de sensores dinamicamente.

```python
# 1. Definição do dicionário do alimentador com seus atributos iniciais
alimentador = {
    "tag": "AL-01",
    "subestacao": "SE-CENTRAL",
    "tensao_kv": 13.8,
    "corrente_a": 142.5,
    "disjuntor_fechado": True
}

# 2. Consulta segura de uma chave inexistente usando .get() com valor padrão (fallback)
# Evita o lançamento de um KeyError catastrófico caso o sensor de temperatura esteja offline
temp_consultada = alimentador.get("temperatura_oleo_c", 25.0)
print(f"[STATUS] Temperatura consultada de forma segura: {temp_consultada} °C (Uso de valor padrão)\n")

# 3. Atualização do dicionário com novos dados de telemetria usando .update()
alimentador.update({
    "temperatura_oleo_c": 78.2,
    "alarme": "ALERTA_TERMICO"
})

# 4. Exibição formatada dos pares chave-valor iterando sobre .items()
print("=== TELEMETRIA DO ALIMENTADOR ATUALIZADA ===")
for chave, valor in alimentador.items():
    print(f"{chave.upper():<20}: {valor}")
```

## Explicação do Código para a Apresentação

* **`alimentador = {...}`**: Criamos uma estrutura de dicionário em Python para modelagem semântica. Ao contrário de uma lista ou tupla, acessamos as informações do alimentador diretamente por chaves textuais descritivas, eliminando erros de indexação numérica.
* **`alimentador.get("temperatura_oleo_c", 25.0)`**: Acessa a temperatura de forma segura. Se a chave `
