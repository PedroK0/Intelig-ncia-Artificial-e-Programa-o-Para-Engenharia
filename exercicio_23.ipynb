```markdown
# Exercício 23: Cálculo Matricial de Perdas Ôhmicas Horárias via Broadcasting

Nesta seção, calculamos as perdas por efeito Joule ($P = R \cdot I^2$) para quatro patamares horários distintos ao longo das três fases. Utilizaremos o mecanismo de **Broadcasting do NumPy** para operar a matriz de correntes $(4 \times 3)$ diretamente com o vetor unidimensional de resistências de fase $(3,)$, executando a multiplicação de maneira eficiente e sem estruturas de repetição em Python.
```

````python
import numpy as np

# 1. Matriz de correntes (4 patamares horários x 3 fases) em Ampères
i_matriz = np.array([
    [80.0, 82.0, 81.0],
    [95.0, 94.0, 97.0],
    [120.0, 118.0, 122.0],
    [60.0, 59.0, 61.0]
])

# 2. Vetor unidimensional de resistências por fase (A, B, C) em Ohms
r_fases = np.array([0.15, 0.15, 0.15])

# 3. Cálculo das perdas por efeito Joule de cada fase em cada patamar (P = I² * R)
# O NumPy aplica automaticamente o vetor r_fases (dimensão 3,) para cada linha da matriz i_matriz (dimensão 4x3) via Broadcasting
p_perdas_matriz = (i_matriz ** 2) * r_fases

# 4. Cálculo da perda total trifásica (em Watts) para cada um dos patamares somando ao longo do eixo das colunas (axis=1)
perdas_trifasicas_por_hora = np.sum(p_perdas_matriz, axis=1)

# --- Exibição formatada para apresentação ---
print("=== ANÁLISE DE PERDAS POR EFEITO JOULE ===\n")
print("Matriz de Perdas por Fase e Patamar (W):")
print(p_perdas_matriz)
print("\n=== PERDAS TRIFÁSICAS TOTAIS POR PATAMAR HORÁRIO ===")
for hora, perdas in enumerate(perdas_trifasicas_por_hora, start=1):
    print(f"Hora/Patamar {hora}: {perdas:.2f} W")

print("\nSaída Esperada:")
print(f"Perdas trifásicas por hora: {list(np.round(perdas_trifasicas_por_hora, 2))} W")
````

```markdown
## Explicação do Código para a Apresentação

* **`i_matriz ** 2`**: Eleva ao quadrado todos os elementos de corrente de forma instantânea e vetorizada.
* **`Broadcasting ((4x3) * (3,))`**: O NumPy alinha a última dimensão da matriz `i_matriz` (que possui tamanho 3) com o vetor `r_fases` (tamanho 3). A operação de multiplicação é realizada de maneira direta para cada uma das linhas sem replicar fisicamente o vetor em memória, otimizando o consumo de recursos.
* **`np.sum(p_perdas_matriz, axis=1)`**: Somar ao longo de `axis=1` (colunas) reduz a dimensão da matriz, somando as contribuições das perdas das fases A, B e C para cada patamar horário individualmente, fornecendo o resultado total em Watts por hora.
```
