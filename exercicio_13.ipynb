```python
# Exercício 13: Fatiamento Temporal e Indicadores de Carga de Alimentador

# Leituras horárias de demanda ativa (kW) de 00h às 23h
demanda_diaria = [
    42.0, 39.5, 38.0, 37.5, 41.0, 56.0, 72.0, 88.5, 94.0, 96.0, 91.5, 89.0,
    85.0, 88.0, 92.5, 98.0, 115.0, 142.0, 155.0, 148.5, 125.0, 98.0, 68.0, 50.0
]

# 1. Extrair o subvetor correspondente ao turno da madrugada (00h às 05h inclusive -> índices de 0 a 5)
turno_madrugada = demanda_diaria[0:6]

# 2. Extrair o subvetor com amostragem bi-horária (a cada 2 horas)
amostragem_bihoraria = demanda_diaria[::2]

# 3. Obter a demanda máxima e mínima
P_max = max(demanda_diaria)
P_min = min(demanda_diaria)

# 4. Calcular o Fator de Carga diário (F_c = P_média / P_máxima)
P_media = sum(demanda_diaria) / len(demanda_diaria)
F_c = P_media / P_max

# Exibição dos resultados
print(f"Subvetor Turno da Madrugada (00h - 05h): {turno_madrugada}")
print(f"Subvetor Amostragem Bi-horária: {amostragem_bihoraria}")
print(f"Demanda Máxima (P_máx): {P_max:.1f} kW")
print(f"Demanda Mínima (P_mín): {P_min:.1f} kW")
print(f"Fator de Carga Diário (F_c): {F_c:.3f}")
```
