# Exercício 18: Algoritmo de Varredura Multicanal e Cálculo de Desbalanceamento de Fases

Este algoritmo realiza o cálculo de desbalanceamento percentual de correntes elétricas para três alimentadores industriais e os classifica dinamicamente de acordo com o limite de tolerância de 15%.
```

````python
# 1. Criação do banco de dados simulado dos alimentadores industriais
alimentadores = [
    {"nome": "Alimentador 1", "correntes": [115.0, 114.0, 116.0]},
    {"nome": "Alimentador 2", "correntes": [140.0, 138.0, 142.0]},
    {"nome": "Alimentador 3", "correntes": [90.0, 125.0, 88.0]}
]

# 2. Iteração multicanal para processar cada alimentador individualmente
for alt in alimentadores:
    correntes = alt["correntes"]

    # Cálculo da corrente média (I_médio)
    i_medio = sum(correntes) / len(correntes)

    # Cálculo do desvio máximo absoluto em relação à média: max(|I_fase - I_médio|)
    desvio_max = max(abs(i - i_medio) for i in correntes)

    # Cálculo do percentual de desbalanceamento
    desbal_pct = (desvio_max / i_medio) * 100

    # Armazenamento dos resultados calculados no dicionário de dados
    alt["desbalanceamento_pct"] = round(desbal_pct, 2)
    alt["i_medio"] = round(i_medio, 2)

    # Classificação de segurança baseada no limite normativo de 15%
    if desbal_pct > 15.0:
        alt["status"] = "ALERTA_DESBALANCEAMENTO"
    else:
        alt["status"] = "CONFORME"

# 3. Exibição formatada dos resultados para análise diagnóstica
print("=== DIAGNÓSTICO DE DESBALANCEAMENTO DE CORRENTE ===\n")
for alt in alimentadores:
    print(f"{alt['nome']}:")
    print(f"  - Correntes (A): {alt['correntes']}")
    print(f"  - Corrente Média: {alt['i_medio']} A")
    print(f"  - Desbalanceamento: {alt['desbalanceamento_pct']:.2f}%")
    print(f"  - Status: {alt['status']}\n")
````

```markdown
## Explicação do Código para a Apresentação

* **`alimentadores = [...]`**: Definimos uma lista contendo dicionários estruturados. Cada dicionário representa um ativo real de telemetria multicanal com suas respectivas leituras de fases A, B e C.
* **`for alt in alimentadores:`**: Iniciamos uma varredura sequencial para analisar as correntes de cada alimentador de forma totalmente automatizada.
* **`i_medio = sum(correntes) / len(correntes)`**: Calcula a média aritmética simples das três correntes de fase coletadas no barramento.
* **`max(abs(i - i_medio) for i in correntes)`**: Descobrimos qual das fases apresenta o maior desvio absoluto em relação à média calculada.
* **`desbal_pct = (desvio_max / i_medio) * 100`**: Aplica a equação física de desbalanceamento nominal para obter a porcentagem de distorção.
* **`if desbal_pct > 15.0:`**: Compara a grandeza física com a restrição operacional de engenharia para aplicar o alarme protetivo ou o selo de conformidade.
